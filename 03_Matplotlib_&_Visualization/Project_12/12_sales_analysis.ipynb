### Data and Output Folder

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

py_path = os.getcwd()

data_folder = os.path.join(py_path, "Data")
output_folder = os.path.join(py_path, "Output")

for folder in [data_folder,output_folder]:
    if not os.path.exists(folder):
        os.mkdir(folder)
        print(f"Created {folder}.")
```

```python
df = pd.read_csv("Data/Sample - Superstore.csv", encoding="windows-1252")

df.shape

df.columns

df["Order Date"] = df["Order Date"].astype("datetime64[ns]")

df.isnull().sum()

category_list = ["Segment","City","Region","Category","Sub-Category"]

df[category_list].nunique()

# print(df["Category"].unique())
# print(df["Sub-Category"].unique())
# print(df["Region"].unique())
# print(df["City"].unique())
# print(df["Segment"].unique())


```

```python

cat_sale = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

print(cat_sale)

Reg_sale = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)
print(Reg_sale)

seg_sale = df.groupby("Segment")["Sales"].sum().sort_values(ascending=False)
print(seg_sale)

sub_cat_sale = df.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False).head(10)
print(sub_cat_sale)
```

```python
fig, ax1 = plt.subplots(figsize=(10, 6))
bars = ax1.bar(cat_sale.index, cat_sale.values, color=['#1f77b4', '#ff7f0e', '#2ca02c'])

# Add value labels on top of bars
ax1.bar_label(bars, padding=3, fmt='${:,.0f}')

# Fix scientific notation on Y-Axis
ax1.yaxis.set_major_formatter(mticker.StrMethodFormatter('{x:,.0f}'))

plt.title("Total Sales by Category")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.savefig("Output/Vertical_bar_Chart.png",dpi=150,bbox_inches='tight')
plt.show()
```

```python
# Create the figure
fig, ax1 = plt.subplots(figsize=(12, 8))

# Note: For horizontal charts, putting the highest value at the TOP 
# often requires plotting the sorted data in reverse or using ax.invert_yaxis()
top_10_sub = sub_cat_sale.head(10).sort_values(ascending=True) 
bars = ax1.barh(top_10_sub.index, top_10_sub.values, color='skyblue')

# Add value labels to the end of bars
ax1.bar_label(bars, padding=5, fmt='${:,.0f}')

# Format X-axis
ax1.xaxis.set_major_formatter(mticker.StrMethodFormatter('{x:,.0f}'))

plt.title("Top 10 Sub-Categories by Sales")
plt.xlabel("Total Sales ($)")
plt.tight_layout()
plt.savefig("Output/Horizontal_bar_Chart.png",dpi=150,bbox_inches='tight')
plt.show()
```

```python
# Create the figure
fig, ax2 = plt.subplots(figsize=(10, 8))

# Explode the 1st slice (West) - check your data order!
# If West is first: (0.1, 0, 0, 0)
explode = (0.1, 0, 0, 0) 

ax2.pie(Reg_sale.values, labels=Reg_sale.index, 
       autopct='%1.1f%%', explode=explode, shadow=True, startangle=140)

plt.title("Sales Contribution by Region")
plt.tight_layout()
plt.savefig("Output/Pie_Chart.png",dpi=150,bbox_inches='tight')
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 8))

top_10_sub = sub_cat_sale.head(10).sort_values(ascending=True) 
bars = ax1.barh(top_10_sub.index, top_10_sub.values, color='skyblue')

# Add value labels to the end of bars
ax1.bar_label(bars, padding=5, fmt='${:,.0f}')

ax2.pie(Reg_sale.values, labels=Reg_sale.index, 
       autopct='%1.1f%%', explode=explode, shadow=True, startangle=140)

plt.title("Sales Contribution by Region")
plt.show()

plt.suptitle("Executive Sales Dashboard", fontsize=16)
plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # Adjust to make room for suptitle
plt.savefig("Output/Subplot_Bar_&_Pie_Chart.png",dpi=150,bbox_inches='tight')
plt.show()
```

```python
import numpy as np

# Prepare data
cat_sale = df.groupby('Category')[['Sales', 'Profit']].sum()
categories = cat_sale.index
x = np.arange(len(categories)) #
width = 0.35 # Width of each bar

fig, ax = plt.subplots(figsize=(12, 7))

# Offset Sales to the left, Profit to the right
ax.bar(x - width/2, cat_sale['Sales'], width, label='Sales', color='royalblue')
ax.bar(x + width/2, cat_sale['Profit'], width, label='Profit', color='seagreen')

# Set the X-ticks to the Category names
ax.set_xticks(x)
ax.set_xticklabels(categories)

ax.legend()
plt.title("Sales vs Profit by Category")
plt.tight_layout()
plt.savefig("Output/Sales_Profit_Chart.png",dpi=150,bbox_inches='tight')
plt.show()
```
